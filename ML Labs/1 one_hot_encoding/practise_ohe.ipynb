```python
data = pd.read_csv("homeprices.csv")
data.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
```

```python
encoded_data = encoder.fit_transform(data[['town']]).toarray()
encoded_data
```

```python
column_names = encoder.get_feature_names_out(['town'])
encoder_df = pd.DataFrame(encoder_data, columns=column_names)
encoder_df
```

```python
final_df = pd.concat([data[['area','price']], encoder_df], axis=1)
final_df.head()
```
